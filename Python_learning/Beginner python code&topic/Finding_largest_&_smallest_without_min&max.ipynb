# Find the Largest and Smallest Number

Write a program that:

Accepts 10 numbers from the user.

Stores them in a list

Finds the largest and smallest number without using max() or min()

Example

Input: [4,7,1,9,3]

Output:
Largest = 9
Smallest = 1

Concepts to use

Lists

Loop

if conditions

```python
user_input = [98,2,3,4,5,6,7,8,9,8,9,22,21,7,7,4]
largest, smallest = largest_smallest(user_input)
print (f'largest = {largest}')
print (f'smallest = {smallest}')
```

```python
def largest_smallest(user_input):
    max_num = user_input[0]
    min_num = user_input[0]
    for i in user_input[1:]:
        
        if i > max_num:
            max_num = i
        elif i < min_num:
            min_num = i
    return max_num,min_num


```
