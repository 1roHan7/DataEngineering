# Reverse Words in a Sentence

Write a Python program that:

Takes a sentence from the user.

Reverses each word individually, but keeps the word order the same.

Example

Input

Hello World Python

Output

olleH dlroW nohtyP

```python
# reversing a string
def reversered_str(user_input):
    reverse_str = []
    split_User_input = user_input.split()
    for i in split_User_input:
        temp = i[::-1] # reverse a string
        #reverse_str = reverse_str + temp +" " # better to use Join
        # seperator.join(list)
        reverse_str.append(temp)

    return (" ".join(reverse_str))
```

```python
# reversing a string with capital letter
def reversered_str_with_cap(user_input):
    reverse_str = []
    caps_list = []
    split_User_input = user_input.split()
    for i in split_User_input:
        temp = i[::-1] # reverse a string
        #reverse_str = reverse_str + temp +" " # better to use Join
        # seperator.join(list)
        reverse_str.append(temp)
    for i in reverse_str:
        caps_list.append(i.capitalize())

    return (" ".join(caps_list))
```
