```python
import pandas as pd
import zipfile
```

```python
import os
zip_path = '/home/rohan/DataEngineering/Python_learning/Project-2-File-to-Database-Loader/Video+Game+Sales.zip'
print(os.path.exists(zip_path))  # Check if the zip file exists
```

```python
with zipfile.ZipFile(zip_path,'r') as z:
    z.extractall('extracted_video_games_sales')
```
