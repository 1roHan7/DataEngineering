- Drop NA value
- Fill NA values
- Imputer Function fill value with Mean median mode

```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder.appName('Practice').getOrCreate()
```

```python
spark_df =spark.read.csv('Book1.csv', header=True,inferSchema=True)
spark_df.show()
```

```python
spark_df.na.drop().show()
# if we dont apply any value in drop it will remove all the rows with Null values 
```

```python
# there are 3 method of drop na 
# how = any(any column has NULL value) / all(only if all columns has NULL value)
# thres = 2 ( so there should be 2 column values for that row with NON NULL values)
# subset = 'Column_name' - Aplly the drop na on a selected row 
spark_df.na.drop(how='all').show()
```

```python
spark_df.na.drop(subset= 'Experience').show()
```

```python
# Fill Na value

spark_df.na.fill(0,[' Age']).show()
```

```python
#Imputer Fucntion
from pyspark.ml.feature import Imputer
Imputer = Imputer(
    inputCols=[' Age','Experience'],
    outputCols=["{}_imputed".format(c) for c in [' Age','Experience']]
).setStrategy('median') # can use Mean median and mode
```

```python
#add imputation col as df
Imputer.fit(spark_df).transform(spark_df).show()
```
